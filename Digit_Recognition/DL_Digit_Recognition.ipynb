```python
(x_train,y_train),(x_test,y_test) = keras.datasets.mnist.load_data() # Loading and splitting Dataset
```

```python
x_train.shape # finding the shape of x_train
```

```python
x_train = x_train/255 # Normalization
```

```python
y_train.shape
```

```python
x_test.shape
```

```python
x_test = x_test/255
```

```python
plt.imshow(x_train[0],cmap = "gray") # showing an image.
plt.show()
```

```python
plt.imshow(x_train[5],cmap="gray")
plt.axis("off")
plt.show()
```

```python
model = keras.Sequential([keras.Input(shape=(28,28)), # Input layer
                          keras.layers.Flatten(), # Flatten Image
                          keras.layers.Dense(128, activation = "relu"), # Hidden layer
                          keras.layers.Dense(10,activation="softmax")]) # Output layer
```

```python
model.compile(optimizer = "adam",loss = "sparse_categorical_crossentropy",metrics = ["accuracy"])
```

```python
model.fit(x_train,y_train,epochs = 10,batch_size = 32) 
```

```python
model.evaluate(x_test,y_test)
```

```python
user_prediction=model.predict(x_train[20].reshape(1,28,28,1))
user_prediction
```
