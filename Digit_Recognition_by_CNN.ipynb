```python
import tensorflow as tf # Library to make Deep Learning models
```

```python
import numpy as np  # Keras - module of tensorflow library
```

```python
from tensorflow import keras # Library for Visulisation
```

```python
import matplotlib.pyplot as plt # library used for arrays
```

```python
(x_train,y_train),(x_test,y_test) = keras.datasets.mnist.load_data() # Loading and splitting Dataset
```

```python
x_train = x_train/255 # Normalization
```

```python
x_train.shape # finding the shape of x_train
```

```python
x_train[0][0]
```

```python
x_test = x_test/255
```

```python
x_test.shape
```

```python
x_test[0][0]
```

```python
plt.imshow(x_train[0],cmap="grey")
plt.axis("off")
plt.show()
```

```python
model = tf.keras.Sequential()
from tensorflow.keras import layers
from tensorflow.keras.layers import MaxPooling2D
```

```python
# Input Layer
model.add(layers.Conv2D(32,(3,3),activation="relu",input_shape=(28,28,1)))
model.add(MaxPooling2D(2,2)) 
# Hidden Layer
model.add(layers.Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2))
model.add(layers.Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2)) 
model.add(layers.Flatten())
model.add(layers.Dense(64,activation="relu"))
# Output Layer
model.add(layers.Dense(10,activation="softmax"))
```

```python
model.compile(optimizer= "adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
model.fit(x_train,y_train,epochs=15)
```

```python
model.evaluate(x_test,y_test)
```

```python
user_prediction=model.predict(x_train[20].reshape(1,28,28,1))
user_prediction
```
