```python
import tensorflow as tf
```

```python
train_path = r"C:\Users\HP\Downloads\Genders_Recognition\Train\Train"
```

```python
test_path = r"C:\Users\HP\Downloads\Genders_Recognition\Test\Test"
```

```python
from tensorflow.keras.utils import image_dataset_from_directory
```

```python
image_size = 128
batch_size = 32
```

```python
training_data = image_dataset_from_directory(
                train_path,
                image_size=(image_size,image_size),
                batch_size=32,
                label_mode="int") # image size must be same for all images, batch_size ->for training only 32 images processed in one time.
```

```python
valid_data = image_dataset_from_directory(
             test_path,
             image_size=(image_size,image_size), 
             batch_size=32,
             label_mode="int") # image size must be same for all images, batch_size ->for training only 32 images processed in one time.)
```

```python
class_name= training_data.class_names
```

```python
print(class_name)
```

```python
normalisation=tf.keras.layers.Rescaling(
              1./255
)
```

```python
training_data = training_data.map(
    lambda x,y:(normalisation(x),y
))
```

```python
valid_data = valid_data.map(
    lambda x,y:(normalisation(x),y
))
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras import layers
```

```python
from tensorflow.keras.layers import Conv2D,MaxPooling2D,Flatten,Dense
```

```python
model = tf.keras.Sequential()
```

```python
# Input Layer
model.add(tf.keras.Input(shape = (128,128,3)))

# Hidden Layer
model.add(MaxPooling2D(2,2))
model.add(layers.Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2))
model.add(layers.Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2))
model.add(layers.Flatten())
model.add(layers.Dense(64,activation="relu"))

# Output Layer
model.add(layers.Dense(1,activation="sigmoid"))
```

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
model.summary()
```

```python
model.fit(training_data,epochs=10,validation_data=valid_data)
```
